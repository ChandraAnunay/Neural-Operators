DeepONets stand for 'Deep Operator Networks'. We have been learning Physics Informed Neural Networks (PINNs) for some time now. The major distinction between the two is that while the PINNs learn the function solution to a single ODE/PDE, DeepONets allow us to learn the relevant operators defined for the ODE/PDE.

# 1. Understanding Operator Networks.

The most famous example is of the operator that maps the input/source function to the ODE/PDE solution. Consider the heat equation for example :

$$ u_t = \alpha u_{xx} $$

with appropriate fixed boundary conditions. Also assume the diffusion term $\alpha$ to be constant for now. A PINN can solve this PDE for a given initial condition :

$$ u(x, 0) = a(x) $$

If we train a PINN model for an initial condition function, $a_1(x)$, it shall work good to estimate the solutions, but it can't provide the solutions for a different initial condition, say $a_2(x)$. If we want it, we need to train the PINN model again on the new initial condition.

What a DeepONet aims to do is learn an operator that can provide solutions $u(x,t)$ for a different set of initial conditions $a_i(x)$.

# 2. DeepONets Architecture.

DeepONets typically contain two neural networks :

i. Branch Net

ii. Truck Net

These two networks act on different items and thus help producing the relevant solution function.

### 2.1. Branch Net.

This network takes up the input/source function $a(x)$ and return an output in a learnt latent space. To understand this, first note that we can send in the function $a(x)$ in a continuous sense. Instead we send in the discretized version. This is achieved by evaluating the function $a(x)$ at different locations $x_1, x_2, ..., x_m$ in the domain. These points are called the fixed 'sensor points'.

So the input is :

$$ \begin{pmatrix}

a(x_1) \\
a(x_2) \\
. \\
. \\
a(x_m) \\

\end{pmatrix} \equiv \begin{pmatrix}

a_1 \\
a_2 \\
. \\
. \\
a_m \\

\end{pmatrix} $$

The Branch Net takes this in and spits out another vector in say a d-dimensional latent space. Thus, 

$$
\text{Branch Net} \left(\begin{pmatrix}
a_1 \\
a_2 \\
. \\
. \\
a_m \\

\end{pmatrix}\right) = \begin{pmatrix}
b_1 \\
b_2 \\
. \\
. \\
b_d \\

\end{pmatrix} \equiv \bold{b} $$

### 2.2. Truck Net.

It does something like the Branch Net, but operates on the point at which the output is required. These points at which output is requires, are called 'query points'. So, suppose we want the output at the point $(x_1,t_1)$. The Trunk Net will take this in, and return a vector in the same dimensional latent space as for Branch Net.

$$ \text{Trunk Net} \left(
\begin{pmatrix}
x\\
t
\end{pmatrix}
\right) = \begin{pmatrix}
t_1\\
t_2\\
. \\
. \\
t_d
\end{pmatrix} \equiv \bold{t} $$

### 2.3. Output.

To get the final prediction, $\tilde{u}(x,t)$ we take the outputs of both the Branch Net and Truck Net, and take a dot product of the two :

$$ \tilde{u}(x,t|a(x)) = \bold{b}(a) \cdot \bold{t}(x,t) = \sum_{i=1}^d b_i(a)t_i(x,t) $$
