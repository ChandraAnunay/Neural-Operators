Previously, we learnt and addressed some of the PDEs using the DeepONet Architecture. Although, the model helped learn the solutions to PDEs and ODEs with different unseen initial conditions, there we some limitations to the Branch and Trunk Networks.

In this notebook, we shall discuss what Fourier Neural Operator (FNOs) architectures are, and how they resolve some of the limitations of DeepONet models. The most important of them being the fact that while DeepONets can only approximate the solutions at fixed query points, thus allowing a limited resolution, the FNOs completely eradicate this problem of fixed query points (or low resolution).

# 1. FNO Architecture.

### 1.1. Motivation.

The solutions of ODEs/PDEs are usually smooth functions. This hints at the idea that most of them could be represented as fourier sums. Further, since they are smooth and non-erratic, we may safely assume as a first guess that the truncated sum of their fourier expansion must be a good approximation to the exact solution. Thus instead of directly applying MLP (Multi-layer Perceptron) in the position space, the model first sends the coordinates to fourier space, apply matrix multiplication with a weigths matrix (to learn fourier coefficients) and then send it back to the position space.

We shall just bluntly give out what happens in the Fourier layers of these model architectures :

$$ v_{l+1} = \sigma(W_lv_l + \mathcal{F}^{-1}[R_l(k)\mathcal{F}[v_l](k)](x)) $$
